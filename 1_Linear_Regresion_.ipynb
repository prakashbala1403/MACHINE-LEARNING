```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dfff=pd.read_csv("mobiles.csv")
dfff
```

```python
dfff.isnull().sum()
```

```python
dfff.dropna(inplace=True)
dfff.isnull().sum()
```

```python

dfff.columns
```

```python
dfff.drop(["mobile_name","extra_storage"],axis=1,inplace=True)
```

```python
dfff
```

```python
a=pd.DataFrame(dfff[["storage"]])
a[['RAM', 'ROM']] = a['storage'].str.split(',', expand=True)

a['RAM'] = a['RAM'].str.strip()
a['ROM'] = a['ROM'].str.strip()
a.drop(columns="storage",axis=1,inplace=True)
print(a)
```

```python
dff=pd.concat([dfff,a],axis=1)
dff.drop("storage",axis=1,inplace=True)
```

```python
dff
```

```python
b=pd.DataFrame(dff["battery"])
b[["MAH","WATTS"]]=b["battery"].str.split("with",expand=True)

b["MAH"]=b["MAH"].str.strip()
b["WATTS"]=b["WATTS"].str.strip()
b
```

```python
df_f=pd.concat([dff,b],axis=1)
df_f.drop("battery",axis=1,inplace=True)
df_f
```

```python
c = pd.DataFrame(df_f["display"])

# Safe split on the first comma only
c[["INCHES","RESOLUTION"]] = c["display"].str.split(",", n=1, expand=True) # split use only 1st comma not two or more

# Assign with column safety
c["INCHES"] = c["INCHES"].str.strip()
c["RESOLUTION"] = c["RESOLUTION"].str.strip() 

c
```

```python
df_final=pd.concat([df_f,c],axis=1)
df_final.drop(columns="display",axis=1,inplace=True)
df_final
```

```python
df_final.columns
```

```python
d=pd.DataFrame(df_final["RESOLUTION"])
d[["Resolution","HZ"]]=d["RESOLUTION"].str.split(",",expand=True)

d["Resolution"]=d["Resolution"].str.strip()
d["HZ"]=d["HZ"].str.strip()
d
df_fi=pd.concat([df_final,d],axis=1)
df_fi.drop(columns="RESOLUTION",axis=1,inplace=True)
df_fi
```

```python
df_fi["RAM"]=df_fi["RAM"].str.replace(r"\s*[a-zA-Z]","",regex=True)
df_fi["ROM"]=df_fi["ROM"].str.replace(r"\s*[a-zA-Z]","",regex=True)  # using regex is very very better
df_fi["MAH"]=df_fi["MAH"].str.replace(r"\s*[a-zA-Z]","",regex=True)
df_fi["WATTS"]=df_fi["WATTS"].str.replace(r"[a-zA-Z]","",regex=True)   # using regex is very very better

df_fi["INCHES"]=df_fi["INCHES"].str.replace("\s*[a-zA-Z]","",regex=True)

df_fi["HZ"]=df_fi["HZ"].str.extract(r"(\d+)")
df_fi.isnull().sum()


df_fi
```

```python
df_fi.isnull().sum()
df_fi.dropna(inplace=True)
df_fi.isnull().sum()
```

```python
df_fi.columns
```

```python
df_fi["RAM"].astype(int)
df_fi["ROM"].astype(float)
df_fi["MAH"].astype(int)

df_fi.isnull().sum()
df_fi['WATTS'] = pd.to_numeric(df_fi['WATTS'], errors='coerce')

df_fi["WATTS"].astype(float)
df_fi["specs_score"].astype(float)
df_fi["INCHES"].astype(float)
df_fi["rating"].astype(float)
df_fi["HZ"].astype(float)
df_fi['RAM'] = pd.to_numeric(df_fi['RAM'], errors='coerce')
df_fi['ROM'] = pd.to_numeric(df_fi['ROM'], errors='coerce')
df_fi['MAH'] = pd.to_numeric(df_fi['MAH'], errors='coerce')
df_fi['HZ'] = pd.to_numeric(df_fi['HZ'], errors='coerce')
df_fi['INCHES'] = pd.to_numeric(df_fi['INCHES'], errors='coerce')


nc=df_fi.select_dtypes(include="number").columns
print(nc)
```

```python
df_fi.isnull().sum()
df_fi.dropna(inplace=True)
df_fi.isnull().sum()
```

```python
df_fi
print(df_fi.shape)
```

```python
nc=df_fi.select_dtypes(include="number").columns
print(nc)

# To find Outliers:
for i in nc:
 Q1=df_fi[i].quantile(0.25)
 Q3=df_fi[i].quantile(0.75)
 IQR=Q3-Q1
 lower=Q1-1.5*IQR
 upper=Q3+1.5*IQR
 q=df_fi[(df_fi[i]<lower)|(df_fi[i]>upper)]  # 0 only shows becoz after removed i checked this so
 w=df_fi[(df_fi[i]>lower)&(df_fi[i]<upper)]  # 20 outliers
print(q.shape)
print(w.shape)
print(df_fi.columns)

```

```python

fig,ax=plt.subplots(4,2,figsize=(15,18))
ax1=np.ravel(ax)
for ind,val in enumerate(nc):
    ax1[ind].boxplot([df_fi[val],w[val]],flierprops={"marker":"o","markersize":6,"color":"blue"},tick_labels=["Outliers", "No_outliers"])
    ax1[ind].set_title(val)             
    ax1[ind].set_xlabel("Dataset")     
    ax1[ind].set_ylabel("Values")      
    ax1[ind].grid(True)
plt.tight_layout()
plt.show()
```

```python
fig,ax=plt.subplots(len(nc),2,figsize=(15,30))
for i,col in enumerate(nc):
    ax[i,0].boxplot(df_fi[col],flierprops={"marker":"o","markersize":8,"color":"blue"},tick_labels=["Outliers"])
    ax[i,0].set_title(col)
    ax[i,1].boxplot(w[col],flierprops={"marker":"o","markersize":8,"color":"blue"},tick_labels=["No_Outliers"])
    ax[i,1].set_title(col)
```

```python
w["price"]=w["price"].str.replace(r"[₹,]","",regex=True)
w["price"]=pd.to_numeric(w["price"],errors="coerce")
w["price"].astype(float)
w["price"].dtype

```

```python
features_col=["connectivity","processor","camera","os","Resolution"]
from sklearn.preprocessing import OneHotEncoder,LabelEncoder
onehot=OneHotEncoder(sparse_output=False,handle_unknown="ignore")
encoder=onehot.fit_transform(w[features_col])
one=pd.DataFrame(encoder,columns=onehot.get_feature_names_out(features_col))
one.index=w.index
df=pd.concat([w.drop(w[features_col],axis=1),one],axis=1)
df.columns.rename
```

```python
# To find correlation :  -----> Feature vs feature corr() AND Feature vs Target corr()

df.corr()
```

```python
# 1.) Feature vs Feature Corr():

feature_only=[i for i in df.columns if i !="price"]  # This is for feature reduced for features selection

corr_matrix=df[feature_only].corr().abs()

mask=np.triu(np.ones(corr_matrix.shape),k=1).astype(bool)
high_corr=corr_matrix.where(mask).stack().reset_index()

high_corr.columns=["Feature_1","Feature_2","Correlation"]
high_corr=high_corr[(high_corr["Correlation"]>0.8)]



to_drop=list(set(high_corr["Feature_1"]))
df_final=df.drop(columns=to_drop)


print(len(to_drop)," High correlation feature dropped",)
print(high_corr.shape,"  That highcorr shape ")
print(df.shape ," our original dataframe shape ")
print(df_final.shape,"  our final dataframe shAPE")
print(len(feature_only))
```

```python
# 2.) Feature vs Target Corr():
features=[i for i in df_final.columns if i!="price"]   # After feature reduced for corr check

important_features=df_final[features].corrwith(df_final["price"]).abs().sort_values(ascending=False)
impo=important_features[(important_features>0.3)]
print(impo)         # This will spread index and values based
print(impo.index," all high corr features")
print(impo.values,"  that values with the target")


#Create that into dataframe  like a -----------------------> feature vs feature corr() we do that see below

impo_df = impo.reset_index()
impo_df.columns = ['Feature', 'Correlation_with_price']
print(impo_df)


```

```python
plt.figure(figsize=(10, 8))
sns.barplot(x=impo.values,y=impo.index)
plt.title("Feature Correlation Matrix")
plt.show()
```

```python

a=[i for i in impo_df["Feature"]]    # use any one both are good
a=impo.index                         # good one
x=df_final[a]             # Values plus name only
y=df_final["price"]       # values only
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
scaler.fit(x_train)
x_train=scaler.transform(x_train)
x_test=scaler.transform(x_test)
```
